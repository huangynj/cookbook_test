# Introduction to COPS

Welcome to the **COPS Cookbook**! This collection of notebooks is designed to help you get started with the **Cloud Optical-array probe Processing Software (COPS)**.

## What is COPS?

COPS is a software package for processing and analyzing data from optical array probes (OAPs) used in cloud microphysics research.

## Getting Started

The current cookbook starts with a reproducible installation check. Future workflow chapters can build from this setup to cover:
- Reading OAP data files
- Visualizing particle images
- Calculating particle size distributions
- Advanced analysis techniques

## Prerequisites

To follow along, you will need:
- A basic understanding of Python
- The `cookbook-dev` environment from `environment.yml`
- A source installation of the COPS package, for example `python -m pip install -e ../COPS`


```python
try:
    import cops
except ImportError as exc:
    raise ImportError(
        "COPS is not installed. Recreate or update the cookbook environment, "
        "install the local COPS source with `python -m pip install -e ../COPS`, "
        "then restart this kernel."
    ) from exc

expected_attrs = ("analysis", "io", "variables")
if not all(hasattr(cops, attr) for attr in expected_attrs):
    raise ImportError(
        "The imported 'cops' package is not the Cloud Optical-array probe "
        "Processing Software. Do not use `pip install cops`; use the "
        "environment.yml dependency instead."
    )

version = getattr(cops, "__version__", "unknown")
location = getattr(cops, "__file__", "unknown")
print(f"COPS version: {version}")
print(f"COPS module: {location}")
```
